```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="G3", bins=10, kde=True)
plt.title("Distribution of Final Mathematics Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x="sex", y="G3")
plt.title("Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Final Grade")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="studytime", y="G3")
plt.title("Study Time vs Final Grade")
plt.xlabel("Weekly Study Time Category")
plt.ylabel("Final Grade")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="absences", y="G3")
plt.title("Absences vs Final Mathematics Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.show()
```

```python
grade_corr = df[["G1", "G2", "G3"]].corr()

sns.heatmap(grade_corr, annot=True, cmap="coolwarm",
            vmin=-1, vmax=1)
plt.title("Correlation Between Student Grades")
plt.show()
```
