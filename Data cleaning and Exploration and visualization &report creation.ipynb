STUDENT ACADEMIC PERFORMANCE ANALYSIS: IDENTIFYING FACTORS INFLUENCING STUDENT ACHIEVEMENT
</h1>

This project aims to analyze student academic performance and identify the key factors that influence student achievement. Using the student dataset, the analysis examines factors such as study time, family background, attendance-related information, and previous grades to understand their relationship with final academic results. Python libraries such as Pandas, Matplotlib, and Seaborn are used for data cleaning, exploratory data analysis, statistical analysis, and visualization. The findings can help identify performance patterns and provide insights that may support improved learning outcomes.

<h1 style="color: purple; text-align: center;">
DATA LOADING & INITIAL OVERVIEW
</h1>


```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\user\OneDrive\Desktop\student-mat.csv",
    sep=";"
)

print(df.head())
print("Dataset shape:", df.shape)

df.info()

print("\nStatistical summary:")
display(df.describe())
```

<h1 style="color: purple; text-align: center;">
DATA PRE-PROCESSING
</h1>

```python
# Check missing values
print(df.isnull().sum())
```

```python
# Check duplicate records
print("Duplicate rows:", df.duplicated().sum())
```

```python
 #Remove duplicates
df = df.drop_duplicates()
```

```python
# Check data types
print(df.dtypes)
```

```python
# Create average grade column
df["Average_Grade"] = df[["G1", "G2", "G3"]].mean(axis=1)
```

```python
# Identify students scoring 10 or above in the final exam
passed_students = df[df["G3"] >= 10]
```

```python
print("Students scoring 10 or above:", len(passed_students))
```

<h1 style="color: purple; text-align: center;">
EXPLORATORY DATA ANALYSIS
</h1>



```python
# Overall academic performance
print(df[["G1", "G2", "G3", "Average_Grade"]].describe())
```

```python
# Average final grade by gender
print(df.groupby("sex")["G3"].mean())
```

```python
# Average final grade by study time
print(df.groupby("studytime")["G3"].mean())
```

```python
# Average final grade by parental education
print(df.groupby("Medu")["G3"].mean())
```

```python
# Relationship between previous and final grades
print(df[["G1", "G2", "G3"]].corr())
```

```python
# Average final grade by school
print(df.groupby("school")["G3"].mean())
```

<h1 style="color: purple; text-align: center;">
PIVOT TABLE
</h1>


The pivot table summarizes the average final mathematics grade (G3) by gender. It allows us to compare average academic performance between gender groups. The results describe associations in this dataset and do not establish causation.

```python
# Pivot table: Average grades by gender
pivot_table = df.pivot_table(
    values="G3",
    index="sex",
    aggfunc="mean"
)

print("Average final grade by gender:")
display(pivot_table)
```

<h1 style="color: purple; text-align: center;">
DATA VISUALIZATION
</h1>

```python
df = pd.read_csv(r"C:\Users\user\OneDrive\Desktop\student-mat.csv", sep=";")
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="G3", bins=20, kde=True, color="mediumpurple")

plt.title("Distribution of Students' Final Mathematics Grades", fontsize=14)
plt.xlabel("Final Grade (G3)", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.xticks(range(0, 21, 2))
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```


### Interpretation — Distribution of Final Grades

The histogram shows the distribution of students' final mathematics grades (G3). It helps identify the grade ranges where students are concentrated and whether the distribution is spread out or clustered around particular values. The KDE curve provides a smoothed view of the distribution.


```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.countplot(data=df, x="G3", color="cornflowerblue")

plt.title("Number of Students by Final Mathematics Grade", fontsize=14)
plt.xlabel("Final Grade (G3)", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.xticks(range(0, 21))
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```


### Interpretation — Number of Students by Final Grade

This bar chart shows how many students achieved each final mathematics grade (G3). It helps identify the most frequent grades and compare the number of students across grade values. Grades with taller bars were achieved by more students than grades with shorter bars.


```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="sex", y="G3", hue="sex", legend=False, errorbar=None,
            palette={"F": "lightcoral", "M": "cornflowerblue"})

plt.title("Average Final Mathematics Grade by Gender", fontsize=14)
plt.xlabel("Gender")
plt.ylabel("Average Final Grade (G3)")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="sex", y="G3", hue="sex", legend=False,
            palette={"F": "lightcoral", "M": "cornflowerblue"})

plt.title("Final Mathematics Grade Distribution by Gender", fontsize=14)
plt.xlabel("Gender")
plt.ylabel("Final Grade (G3)")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```


### Interpretation — Final Grade Distribution by Gender

The box plot compares the distribution of final mathematics grades between female and male students. It helps compare their median grades, variability, and potential outliers. The chart should be interpreted using the displayed results rather than assuming one group performed better.


```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="studytime",
    y="G3",
    hue="studytime",
    legend=False,
    errorbar=None,
    color="mediumpurple"
)

plt.title("Average Final Grade by Weekly Study Time", fontsize=14)
plt.xlabel("Weekly Study Time Category")
plt.ylabel("Average Final Grade (G3)")
plt.xticks(
    ticks=[0, 1, 2, 3],
    labels=["Less than 2 hrs", "2–5 hrs", "5–10 hrs", "More than 10 hrs"]
)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```


### Interpretation — Weekly Study Time and Final Grades

This bar chart compares average final mathematics grades across weekly study-time categories. It helps explore whether average grades differ among students who report different amounts of study time. The chart shows an association between study-time category and grades, if present, but does not establish that study time alone causes differences in achievement.


```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="G1",
    y="G3",
    color="teal",
    alpha=0.7,
    s=70
)

plt.title("Relationship Between First Period and Final Grades", fontsize=14)
plt.xlabel("First Period Grade (G1)")
plt.ylabel("Final Grade (G3)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```


### Interpretation — First Period and Final Grades

The scatter plot examines the relationship between first-period grades (G1) and final grades (G3). The pattern of points helps identify whether higher first-period grades tend to occur alongside higher final grades. This is an association and does not, by itself, prove causation.


```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="G2",
    y="G3",
    color="darkblue",
    alpha=0.7,
    s=70
)

plt.title("Relationship Between Second Period and Final Grades", fontsize=14)
plt.xlabel("Second Period Grade (G2)")
plt.ylabel("Final Grade (G3)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```


### Interpretation — Second Period and Final Grades

The scatter plot examines the relationship between second-period grades (G2) and final grades (G3). The distribution of points helps show whether students with higher second-period grades also tend to achieve higher final grades. This relationship indicates association, not proof of causation.


```python

import matplotlib.pyplot as plt
import seaborn as sns

numeric_df = df.select_dtypes(include="number")
correlation_matrix = numeric_df.corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Student Performance Variables", fontsize=14)
plt.tight_layout()
plt.show()
```


### Interpretation — Correlation Heatmap

The heatmap displays correlations among numerical variables in the student dataset. It helps identify variables that move together and relationships that may be worth exploring further. Correlation does not establish causation, and relationships should be interpreted in the context of the data.


```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Medu",
    y="G3",
    hue="Medu",
    legend=False,
    errorbar=None,
    color="seagreen"
)

plt.title("Average Final Grade by Mother's Education Level", fontsize=14)
plt.xlabel("Mother's Education Level (Medu)")
plt.ylabel("Average Final Grade (G3)")
plt.xticks(
    ticks=[0, 1, 2, 3, 4],
    labels=[
        "None",
        "Primary (4th grade)",
        "5th–9th grade",
        "Secondary",
        "Higher education"
    ]
)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```


### Interpretation — Mother's Education and Final Grades

This chart compares average final mathematics grades across categories of mother's education. It helps explore whether student outcomes differ between these groups. Differences in averages should be interpreted cautiously because mother's education is only one of many factors associated with academic performance.


```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="school",
    y="G3",
    hue="school",
    legend=False,
    errorbar=None,
    palette={"GP": "cornflowerblue", "MS": "orange"}
)

plt.title("Average Final Mathematics Grade by School", fontsize=14)
plt.xlabel("School")
plt.ylabel("Average Final Grade (G3)")
plt.xticks(
    ticks=[0, 1],
    labels=["Gabriel Pereira (GP)", "Mousinho da Silveira (MS)"]
)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```


### Interpretation — School and Final Grades

This chart compares average final mathematics grades between Gabriel Pereira (GP) and Mousinho da Silveira (MS). It helps identify differences in average outcomes between the two schools in this dataset. Any observed difference is descriptive and does not establish that the school itself caused the difference.


# CONCLUSION AND RECOMMENDATIONS

## Conclusion

This project analyzed student academic performance using 395 student records and 33 columns from the mathematics student-performance dataset. The analysis included descriptive statistics, data visualizations, group comparisons, and correlation analysis.

The average final grade was 10.42 out of 20. Previous-period grades were positively correlated with final grades, particularly G2, which had a correlation of 0.905 with G3. Average final grades also differed across study-time categories, gender groups, mother's education categories, and schools.

These findings describe patterns in the available dataset. They do not prove that any single factor causes differences in academic performance.

## Recommendations

1. **Monitor earlier grades:** Use first- and second-period results to identify students who may need additional academic support.
2. **Provide targeted support:** Offer tutoring and learning resources to students who are struggling with mathematics.
3. **Encourage effective study habits:** Help students develop consistent study routines and monitor their progress.
4. **Consider students individually:** Avoid making assumptions about a student's ability based on gender, family background, or school.
5. **Explore further:** Examine the number of students in each category and investigate additional academic, family, and personal variables before drawing stronger conclusions.
