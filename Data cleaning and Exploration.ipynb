<h1 style="color: purple; text-align: center;">
DATA LOADING & INITIAL OVERVIEW
</h1>


```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\user\OneDrive\Desktop\student-mat.csv",
    sep=";"
)

print(df.head())
print("Dataset shape:", df.shape)

df.info()

print("\nStatistical summary:")
display(df.describe())
```

<h1 style="color: purple; text-align: center;">
DATA PRE-PROCESSING
</h1>

```python
# Check missing values
print(df.isnull().sum())
```

```python
# Check duplicate records
print("Duplicate rows:", df.duplicated().sum())
```

```python
 #Remove duplicates
df = df.drop_duplicates()
```

```python
# Check data types
print(df.dtypes)
```

```python
# Create average grade column
df["Average_Grade"] = df[["G1", "G2", "G3"]].mean(axis=1)
```

```python
# Identify students scoring 10 or above in the final exam
passed_students = df[df["G3"] >= 10]
```

```python
print("Students scoring 10 or above:", len(passed_students))
```

<h1 style="color: purple; text-align: center;">
EXPLORATORY DATA ANALYSIS
</h1>



```python
# Overall academic performance
print(df[["G1", "G2", "G3", "Average_Grade"]].describe())
```

```python
# Average final grade by gender
print(df.groupby("sex")["G3"].mean())
```

```python
# Average final grade by study time
print(df.groupby("studytime")["G3"].mean())
```

```python
# Average final grade by parental education
print(df.groupby("Medu")["G3"].mean())
```

```python
# Relationship between previous and final grades
print(df[["G1", "G2", "G3"]].corr())
```

```python
# Average final grade by school
print(df.groupby("school")["G3"].mean())
```

<h1 style="color: purple; text-align: center;">
PIVOT TABLE
</h1>


The pivot table summarizes the average final mathematics grade (G3) by gender. It allows us to compare average academic performance between gender groups. The results describe associations in this dataset and do not establish causation.

```python
# Pivot table: Average grades by gender
pivot_table = df.pivot_table(
    values="G3",
    index="sex",
    aggfunc="mean"
)

print("Average final grade by gender:")
display(pivot_table)
```
